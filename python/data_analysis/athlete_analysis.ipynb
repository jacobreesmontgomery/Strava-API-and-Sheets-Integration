## DATA ACQUISITION AND CLEANING

### Read in the data from PostgreSQL and set up a Pandas DataFrame

```python
# Import libraries
import sqlalchemy
import pandas as pd

# Connect to the database
from dao.DatabaseService import DatabaseService
db_service = DatabaseService()
session = db_service.get_session()

# Query for all data
from dao.models.activity import Activity

query_result = session.query(Activity).filter(Activity.athlete_id == 41580846)

# Convert to a list of dictionaries
data = [row.__dict__ for row in query_result]

# Remove the SQLAlchemy state object (if it exists)
for d in data:
    d.pop('_sa_instance_state', None)

# Create a Pandas DataFrame
df = pd.DataFrame(data).sort_values(by="full_datetime", ascending=False)

# Print the columns
print(df.columns)

# Display the first row
df.head(1)
```

### Clean out garbage data

```python
# Clean out null values from the df
null_counts = df.isnull().sum()
cols_with_nulls = null_counts[null_counts > 0]
print(f"Null column counts:\n{cols_with_nulls}")

# Replace null values for wkt_type and description with 0 and "", respectively
df.fillna({"wkt_type": 0, "description": ""}, inplace=True)
```

```python
# Confirming we've cleaned out NaN values
null_counts = df.isnull().sum()
cols_with_nulls = len(null_counts[null_counts > 0])
print(f"Null column counts:\n{cols_with_nulls}")
```
